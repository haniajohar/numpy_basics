```python
# Conceptual piecewise transformation
# high peak: I_new = H + alpha * (I - H)
# low peak : I_new = L + alpha * (I - L)
# middle values remain unchanged
```

```python
def soften_peaks(channel, low=20, high=235, alpha=0.5):
 c = channel.astype(np.float32)
 high_part = high + alpha * (c - high)
 low_part = low + alpha * (c - low)
 c = np.where(c > high, high_part, c)
 c = np.where(c < low, low_part, c)
 return np.clip(c, 0, 255).astype(np.uint8)
red_eq = soften_peaks(red)
green_eq = soften_peaks(green)
blue_eq = soften_peaks(blue)
equalized_img = np.dstack((red_eq, green_eq, blue_eq))
```

Questions
1. Why does changing only the red array alter many final colors, not just pure red regions?
Answer: The final color of every pixel is made by combining the R, G, and B values. If we increase red, any pixel that already has some red in it will change — even if it also has green and blue. So mixed colors like yellow (R+G), magenta (R+B), and white (R+G+B) all shift toward red. Only pixels with zero red stay completely unchanged.

2. What happens when a channel value reaches 255 and you add more intensity without clipping?
Answer: Without clipping, the value overflows. Since the array is uint8, it wraps around (255 + 1 becomes 0, 255 + 2 becomes 1, etc.). This causes bright areas to suddenly become dark, which looks like a strange artifact. That is why we use np.clip to keep values in the 0–255 range.

3. Why is the peak-softening exercise different from true histogram equalization?
Answer: True histogram equalization redistributes all intensity values based on the image's cumulative histogram to make the histogram roughly uniform. It changes every pixel value and is based on global statistics. Peak softening here only compresses values above high and below low toward those thresholds using a fixed alpha. Middle values are untouched. It is a simple, local intensity compression for teaching, not a histogram-based equalization.

```python
import numpy as np
import matplotlib.pyplot as plt
img3 = np.zeros((9, 9, 3), dtype=np.uint8)
# Red block
img3[1:4, 1:4, 0] = 255
# Green block
img3[2:6, 4:7, 1] = 255
# Blue block
img3[5:8, 2:5, 2] = 255
plt.imshow(img3)
plt.title("Directly Created 3-Color Image")
plt.axis("off")
plt.show()
print("Shape:", img3.shape)
```

```python
kernel = np.array([
 [ 0, -1, 0],
 [-1, 4, -1],
 [ 0, -1, 0]
], dtype=np.int32)
```

```python
h, w, ch = img3.shape
edges = np.zeros((h, w, ch), dtype=np.float32)
for c in range(ch): # R, G, B
 for i in range(1, h - 1): # image rows
  for j in range(1, w - 1): # image columns
 # Current 3x3 neighborhood
     window = img3[i-1:i+2, j-1:j+2, c]
 # Element-by-element multiplication and summation
     value = np.sum(window * kernel)
     edges[i, j, c] = value
# Edge strength should be non-negative for display
edges = np.abs(edges)
edges = np.clip(edges, 0, 255).astype(np.uint8)
plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.imshow(img3)
plt.title("Original")
plt.axis("off")
plt.subplot(1, 2, 2)
plt.imshow(edges)
plt.title("Manual 3x3 Edge Response")
plt.axis("off")
plt.show()
```

# Questions and Answers

### 1. Why are edge responses strongest where the color/intensity changes rapidly?

Edge responses are strongest where there is a **rapid change in color or intensity** between neighboring pixels. The edge-detection kernel detects these differences and produces a larger response when the change is greater.



### 2. Why does the same kernel produce different responses in the R, G, and B channels?

An RGB image contains three separate channels: **Red (R), Green (G), and Blue (B)**. Each channel has different intensity values. Therefore, applying the same kernel to each channel produces different responses depending on the intensity changes in that particular channel.



### 3. What happens at the image boundary when a complete 3×3 neighborhood is not available?

At the image boundary, a complete **3×3 neighborhood is not available** because pixels outside the image do not exist. In our code, the boundary pixels are skipped because the loop starts from index `1` and stops before the last pixel. Therefore, the outermost pixels are not processed.



### 4. What visual differences do you observe between the original 3-color image and the edge image?

The original RGB image contains **colors, objects, textures, and background details**. The edge image mainly highlights **boundaries and areas with rapid changes in color or intensity**. As a result, the edge image appears mostly dark with bright lines representing the detected edges.
