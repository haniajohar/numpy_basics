Task 6: Reshaping and Indexing Multi-Dimensional NumPy Arrays

Create a 1D NumPy array with 12 elements (1 to 12). Reshape it into a 3x4 matrix. Extract and display the 2x2 sub- matrix located at the center of the array.

```python
import numpy as np

arr = np.arange(1, 13)
matrix = arr.reshape(3, 4)
sub_matrix = matrix[0:2, 1:3]

print("1D Array:", arr)
print("\n3x4 Matrix:")
print(matrix)

print("\n2x2 Sub-matrix:")
print(sub_matrix)
```

Task 7: Boolean Indexing and Filtering with NumPy

Create a 1D NumPy array containing 15 random integer values ranging from 1 to 100. Write a logic to select and display only values that are greater than 50 and even.

```python
import numpy as np
arr = np.random.randint(1, 101, 15)

print("Original Array:")
print(arr)
filtered = arr[(arr > 50) & (arr % 2 == 0)]

print("\nValues greater than 50 and even:")
print(filtered)
```

Task 8: Conditional Row Selection in Pandas

Create a Pandas DataFrame containing sales records (Product, Unit_Price, Quantity_Sold). Add a calculated column
Total_Sales (Unit_Price * Quantity_Sold) and filter rows where Total_Sales exceeds 500.

```python
import pandas as pd
data = {
    "Product": ["Laptop", "Mouse", "Keyboard", "Monitor", "Headphones"],
    "Unit_Price": [800, 25, 50, 200, 75],
    "Quantity_Sold": [2, 10, 5, 3, 8]
}

df = pd.DataFrame(data)
df["Total_Sales"] = df["Unit_Price"] * df["Quantity_Sold"]

print("Sales DataFrame:")
print(df)
filtered_df = df[df["Total_Sales"] > 500]

print("\nRows where Total_Sales exceeds 500:")
print(filtered_df)
```

Task 9: Handling Sorting and Ranking in Pandas

Create a Pandas DataFrame containing a list of 6 students with their GPA and Attendance percentage. Sort the DataFrame first by GPA in descending order, and then by Attendance in ascending order.

```python
import pandas as pd

data = {
    "Student": ["Ali", "Sara", "Hania", "Ahmed", "Ayesha", "Usman"],
    "GPA": [3.2, 3.8, 3.8, 3.1, 3.5, 3.8],
    "Attendance": [85, 90, 80, 95, 88, 85]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)
sorted_df = df.sort_values(
    by=["GPA", "Attendance"],
    ascending=[False, True]
)

print("\nSorted DataFrame:")
print(sorted_df)
```


Task 10: Combining DataFrames via Concatenation and Merging

Create two DataFrames representing store sales from Region A and Region B. Concatenate them vertically, calculate total sales for each region, and merge with a region metadata table.

```python
import pandas as pd

region_a = pd.DataFrame({
    "Product": ["Laptop", "Mouse", "Keyboard"],
    "Unit_Price": [800, 25, 50],
    "Quantity_Sold": [2, 10, 5],
    "Region": ["A", "A", "A"]
})

region_b = pd.DataFrame({
    "Product": ["Laptop", "Monitor", "Headphones"],
    "Unit_Price": [800, 200, 75],
    "Quantity_Sold": [1, 3, 8],
    "Region": ["B", "B", "B"]
})
combined = pd.concat([region_a, region_b], ignore_index=True)
combined["Total_Sales"] = combined["Unit_Price"] * combined["Quantity_Sold"]

print("Combined Sales Data:")
print(combined)
region_totals = combined.groupby("Region")["Total_Sales"].sum().reset_index()

print("\nTotal Sales for Each Region:")
print(region_totals)

region_metadata = pd.DataFrame({
    "Region": ["A", "B"],
    "Manager": ["Ali", "Sara"],
    "City": ["Islamabad", "Rawalpindi"]
})
merged = pd.merge(region_totals, region_metadata, on="Region")

print("\nMerged Data:")
print(merged)
```
