```python
import math


def Menu():
    print("\n----- CALCULATOR MENU -----")
    print("1. Addition")
    print("2. Subtraction")
    print("3. Multiplication")
    print("4. Division")
    print("5. Factorial")
    print("6. x_power_y")
    print("7. Log")
    print("8. Natural Log (ln)")
    print("9. Quit")

    choice = int(input("Enter your choice: "))
    return choice


def addition():
    a = float(input("Enter first number: "))
    b = float(input("Enter second number: "))
    print("Result =", a + b)


def subtraction():
    a = float(input("Enter first number: "))
    b = float(input("Enter second number: "))
    print("Result =", a - b)


def multiplication():
    a = float(input("Enter first number: "))
    b = float(input("Enter second number: "))
    print("Result =", a * b)


def division():
    a = float(input("Enter first number: "))
    b = float(input("Enter second number: "))

    if b == 0:
        print("Cannot divide by zero.")
    else:
        print("Result =", a / b)


def factorial():
    n = int(input("Enter a number: "))

    if n < 0:
        print("Factorial is not defined for negative numbers.")
    else:
        print("Result =", math.factorial(n))


def x_power_y():
    x = float(input("Enter x: "))
    y = float(input("Enter y: "))

    print("Result =", x ** y)


def log():
    x = float(input("Enter a number: "))

    if x <= 0:
        print("Log is not defined for zero or negative numbers.")
    else:
        print("Result =", math.log10(x))


def ln():
    x = float(input("Enter a number: "))

    if x <= 0:
        print("Natural log is not defined for zero or negative numbers.")
    else:
        print("Result =", math.log(x))


while True:
    choice = Menu()

    if choice == 1:
        addition()
    elif choice == 2:
        subtraction()
    elif choice == 3:
        multiplication()
    elif choice == 4:
        division()
    elif choice == 5:
        factorial()
    elif choice == 6:
        x_power_y()
    elif choice == 7:
        log()
    elif choice == 8:
        ln()
    elif choice == 9:
        print("Program terminated.")
        break
    else:
        print("Invalid choice. Please try again.")
```

```python
%%writefile basic.py

class basic_calc:

    def __init__(self, a, b):
        self.a = a
        self.b = b

    def addition(self):
        return self.a + self.b

    def subtraction(self):
        return self.a - self.b

    def multiplication(self):
        return self.a * self.b

    def division(self):
        if self.b == 0:
            return "Cannot divide by zero"
        return self.a / self.b
```

```python
import basic
import math


class s_calc(basic.basic_calc):

    def factorial(self):
        return math.factorial(self.a)

    def x_power_y(self):
        return self.a ** self.b

    def log(self):
        return math.log10(self.a)

    def ln(self):
        return math.log(self.a)


x = s_calc(5, 2)

print("Addition =", x.addition())
print("Subtraction =", x.subtraction())
print("Multiplication =", x.multiplication())
print("Division =", x.division())
print("Factorial =", x.factorial())
print("x Power y =", x.x_power_y())
print("Log =", x.log())
print("Natural Log =", x.ln())
```

** Lab 1-E: First Steps in Image Processing with Python**

```python
# Digital Image Processing with Python
# Install once if needed: pip install pillow numpy matplotlib
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt
# 1. Acquire / load an image
image_path = input("Enter the image file path: ")
image = Image.open(image_path).convert("RGB")
img = np.array(image)
print("Image shape:", img.shape)
print("Data type:", img.dtype)
# 2. Extract the three color-channel arrays
red = img[:, :, 0].copy()
green = img[:, :, 1].copy()
blue = img[:, :, 2].copy()
print("Red array shape:", red.shape)
print("Green array shape:", green.shape)
print("Blue array shape:", blue.shape)
# 3. Recombine the original channels
original = np.stack([red, green, blue], axis=2)
plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.imshow(original)
plt.title("Original image")
plt.axis("off")
# 4. Change selected pixel values
# Increase red in a rectangular region
red[50:200, 50:200] = 255
# Reduce green in another region
green[100:250, 100:250] = (green[100:250, 100:250] * 0.40).astype(np.uint8)
# Increase blue in another region (with clipping)
blue[150:300, 150:300] = np.clip(
 blue[150:300, 150:300].astype(np.int16) + 80, 0, 255
).astype(np.uint8)
# 5. Recombine the modified channels
modified = np.stack([red, green, blue], axis=2)
plt.subplot(1, 2, 2)
plt.imshow(modified)
plt.title("Modified image")
plt.axis("off")
plt.tight_layout()
plt.show()
```

**Lab 1-E - Understanding the RGB Arrays**

```python
# Optional channel inspection
plt.figure(figsize=(12, 3))
plt.subplot(1, 3, 1)
plt.imshow(red, cmap="gray")
plt.title("Red channel")
plt.axis("off")
plt.subplot(1, 3, 2)
plt.imshow(green, cmap="gray")
plt.title("Green channel")
plt.axis("off")
plt.subplot(1, 3, 3)
plt.imshow(blue, cmap="gray")
plt.title("Blue channel")
plt.axis("off")
plt.tight_layout()
plt.show()
```

```python
# Make copies so the original channel arrays remain unchanged
red_c = red.copy()
green_c = green.copy()
blue_c = blue.copy()

# Change larger/different regions
red_c[50:250, 50:250] = 255

green_c[200:400, 200:400] = 0

blue_c[300:500, 300:500] = 255

# Recombine
modified_c = np.stack([red_c, green_c, blue_c], axis=2)

# Display
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(original)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(modified_c)
plt.title("Task 1 - Changed Regions")
plt.axis("off")

plt.tight_layout()
plt.show()
```

I changed the coordinates of the selected regions and made them larger. The modified regions appeared in different colors and positions because the RGB values of those pixels were changed.

```python
red_c = red.copy()
green_c = green.copy()
blue_c = blue.copy()

# Set a region of the Green channel to 0
green_c[100:300, 100:300] = 0

# Recombine
modified_c2 = np.stack([red_c, green_c, blue_c], axis=2)

# Display
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(original)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(modified_c2)
plt.title("Task 2 - Green Channel Changed")
plt.axis("off")

plt.tight_layout()
plt.show()
```

Only the Green channel was modified by setting its selected region to zero. The affected region changed color because the green component was removed while the Red and Blue components remained unchanged.

```python
red_c = red.copy()
green_c = green.copy()
blue_c = blue.copy()

# Reduce Blue intensity
blue_c[100:300, 100:300] = (
    blue_c[100:300, 100:300] * 0.3
).astype(np.uint8)

modified_blue = np.stack([red_c, green_c, blue_c], axis=2)

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(original)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(modified_blue)
plt.title("Blue Channel Reduced")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
red_c = red.copy()
green_c = green.copy()
blue_c = blue.copy()

# Increase Red
red_c[100:300, 100:300] = 255

# Reduce Green
green_c[100:300, 100:300] = 0

# Keep Blue unchanged

modified_c3 = np.stack([red_c, green_c, blue_c], axis=2)

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(original)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(modified_c3)
plt.title("Task 3 - Red Increased + Green Reduced")
plt.axis("off")

plt.tight_layout()
plt.show()
```

Increasing the Red channel while reducing the Green channel changed the selected region toward red/purple. This demonstrates that the final color depends on the combination of the three RGB channel values.

A color image stores three values at every pixel: Red, Green, and Blue. Changing any one of these values changes the color of that pixel. When the three modified channel arrays are recombined, the new RGB values are interpreted by the display as the new color.

```python
# Convert image to floating point for matrix calculations
img_float = original.astype(np.float32)

# Get image dimensions
height, width, channels = img_float.shape

print("Image shape:", img_float.shape)
```

```python
matrix_red = np.array([
    [1.5, 0, 0],
    [0,   1, 0],
    [0,   0, 1]
])

# Apply matrix to every RGB pixel
result_red = img_float @ matrix_red.T

# Keep values between 0 and 255
result_red = np.clip(result_red, 0, 255).astype(np.uint8)

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(original)
plt.title("Original")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(result_red)
plt.title("Red Emphasis")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
matrix_green = np.array([
    [1, 0,   0],
    [0, 1.5, 0],
    [0, 0,   1]
])

result_green = img_float @ matrix_green.T

result_green = np.clip(result_green, 0, 255).astype(np.uint8)

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(original)
plt.title("Original")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(result_green)
plt.title("Green Emphasis")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
matrix_mix = np.array([
    [1.0, 0.2, 0.2],
    [0.2, 1.0, 0.2],
    [0.2, 0.2, 1.0]
])

result_mix = img_float @ matrix_mix.T

result_mix = np.clip(result_mix, 0, 255).astype(np.uint8)

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(original)
plt.title("Original")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(result_mix)
plt.title("RGB Channel Mixing")
plt.axis("off")

plt.tight_layout()
plt.show()
```

### Reflection
In this experiment, I learned that a digital RGB image can be represented as numerical arrays containing Red, Green, and Blue values. Changing the values of these arrays directly changes the visible color of selected pixels or regions. Increasing or decreasing one channel produced different color effects, while changing multiple channels produced combined color changes. I also applied 3 × 3 RGB transformation matrices to mix or emphasize color channels. The results showed that numerical transformations of RGB values can produce noticeable visual changes in the final image. This demonstrates how image processing operations can be understood as mathematical transformations of pixel data.
